```python
from deepeval import evaluate
from deepeval.metrics import FaithfulnessMetric
from deepeval.models import OpenAIModel
from deepeval.test_case import LLMTestCase

# 1. Define a mock test case from your Notes Manager
test_case = LLMTestCase(
    input="Generate a 3-question multiple-choice quiz based on these notes.",
    actual_output="""{
        "quizzes": [
          {"question": "What color is chlorophyll?", "options": ["Red", "Green", "Blue"], "answer": "Green"}
        ]
    }""",
    retrieval_context=["Notes: Chlorophyll is a green pigment found in plants that absorbs light for photosynthesis."]
)

# 2. Define the metrics
model = OpenAIModel(
    model="gpt-5-nano",
)

faithfulness_metric = FaithfulnessMetric(threshold=0.7, model=model)
```

```python
from pathlib import Path
import pandas as pd

CSV_PATH = Path(
    r"D:\llm-projects\projects\study-notes-manager\backend\tests\golden_evaluation_dataset_fixed.csv"
)

df = pd.read_csv(CSV_PATH, keep_default_na=False)

required_columns = {"note_content", "expected_output", "metadata"}
missing = required_columns - set(df.columns)
if missing:
    raise ValueError(f"Missing CSV columns: {missing}")

invalid_input = df[
    df["note_content"].astype(str).str.strip().eq("")
]

if not invalid_input.empty:
    print(
        "Skipping rows with empty note_content:",
        ", ".join(invalid_input["test_case_id"].astype(str))
    )
    df = df.loc[~df.index.isin(invalid_input.index)].copy()

df.to_csv(CSV_PATH, index=False)

dataset = EvaluationDataset()

dataset.add_goldens_from_csv_file(
    file_path=str(CSV_PATH),
    input_col_name="note_content",
    expected_output_col_name="expected_output",
    additional_metadata_col_name="metadata",
)

print(f"Loaded {len(dataset.goldens)} goldens successfully.")
```
